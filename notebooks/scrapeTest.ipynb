Scrape all Titles from bookstoscrape


```python
import requests
from bs4 import BeautifulSoup as bs
from urllib.parse import urljoin
```

```python
#Load the page
url = 'https://books.toscrape.com'

r = requests.get(url)

if r.status_code == 200:
    #Parse the page
    soup = bs(r.content, 'html.parser')

    #Find all the books on the page
    #All the books are contained in <h3> tags, so we can use the find_all method to get all the <h3> tags
    books = soup.find_all('h3')

    #Print the titles of the books
    for book in books:
        print(book.a['title'])
else:
    print(f'Failed to retrieve the page. Status code: {r.status_code}')
```

```python
if r.status_code == 200:
    #Parse the page
    soup = bs(r.content, 'html.parser')

    #Find all the books on the page
    #All the books are contained in <h3> tags, so we can use the find_all method to get all the <h3> tags
    books = soup.find_all('h3')

    #Print the titles and product descriptions from each book page
    for book in books:
        print(book.a['title'])
        new_link = book.a['href']  # Get the link to the book
        book_response = requests.get(f'https://books.toscrape.com/{new_link}')
        soup2 = bs(book_response.content, 'html.parser')  # Load the book page
        description = soup2.select_one('#product_description + p')
        if description:
            print(description.get_text(' ', strip=True))
        
else:
    print(f'Failed to retrieve the page. Status code: {r.status_code}')
```

```python
category_name = 'Fiction'
base_url = 'https://books.toscrape.com/'

homepage_response = requests.get(base_url)
homepage_soup = bs(homepage_response.content, 'html.parser')
category_link = next(
    (
        link
        for link in homepage_soup.select('ul.nav-list a')
        if link.get_text(strip=True).casefold() == category_name.casefold()
    ),
    None
)

if category_link:
    category_url = urljoin(base_url, category_link['href'])
    print(f'Scraping category: {category_name}')
    print(f'Category URL: {category_url}')

    category_response = requests.get(category_url)
    if category_response.status_code == 200:
        soup = bs(category_response.content, 'html.parser')
        books = soup.find_all('h3')

        for book in books:
            print(book.a['title'])
            book_url = urljoin(category_url, book.a['href'])
            book_response = requests.get(book_url)
            soup2 = bs(book_response.content, 'html.parser')
            description = soup2.select_one('#product_description + p')
            img = soup2.select_one('.item.active img')
            if description:
                print(description.get_text(' ', strip=True))
            if img:
                print(f'Image URL: {urljoin(book_url, img["src"])}')
    else:
        print(f'Failed to retrieve the category page. Status code: {category_response.status_code}')
else:
    print(f'Category not found: {category_name}')
```

Now we have base code for scraping any specific categorie on the site that we want. Next up is how to work around scraping the new pages

When looking at the sites i noticed that the urls for different pages are just ./page-[number].html


```python
def scrape_category(category_name):
    category_link = next(
        (
            link
            for link in homepage_soup.select("ul.nav-list a")
            if link.get_text(strip=True).casefold()
            == category_name.casefold()
        ),
        None,
    )

    if category_link is None:
        print(f"Category not found: {category_name}")
        return

    category_url = urljoin(base_url, category_link["href"])

    while category_url:
        print(f"Scraping page: {category_url}")

        category_response = requests.get(category_url)

        if category_response.status_code != 200:
            print(
                f"Failed to retrieve page. "
                f"Status code: {category_response.status_code}"
            )
            break

        soup = bs(category_response.content, "html.parser")
        books = soup.find_all("h3")

        for book in books:
            print(book.a["title"])

            book_url = urljoin(category_url, book.a["href"])
            book_response = requests.get(book_url)
            soup2 = bs(book_response.content, "html.parser")

            description = soup2.select_one("#product_description + p")
            img = soup2.select_one(".item.active img")

            if description:
                print(description.get_text(" ", strip=True))

            if img:
                print(f"Image URL: {urljoin(book_url, img['src'])}")

        next_link = soup.select_one("li.next a")

        if next_link:
            category_url = urljoin(category_url, next_link["href"])
        else:
            category_url = None


scrape_category("Nonfiction")
```
